```python
USUARIO = "estudiante"
CONTRASENA = ""

HOST = "34.39.202.205"
PUERTO = 6432
BASE = "retail_dwh"
```

```python
%pip install -q "psycopg[binary]" "SQLAlchemy>=2" pandas

import pandas as pd
from sqlalchemy import create_engine
from sqlalchemy.engine import URL
from sqlalchemy.pool import NullPool

if not CONTRASENA.strip():
    raise SystemExit("Escribe la contraseña en la celda anterior y vuelve a ejecutarla.")

engine = create_engine(
    URL.create(
        "postgresql+psycopg",
        username=USUARIO,
        password=CONTRASENA,
        host=HOST,
        port=PUERTO,
        database=BASE,
    ),
    poolclass=NullPool,
    connect_args={
        "sslmode": "require",
        "prepare_threshold": None,
        "connect_timeout": 20,
    },
)

def consultar(sql):
    return pd.read_sql(sql, engine)
```

```python
consultar("""
SELECT *
FROM curso.ventas
LIMIT 10
""")
```
